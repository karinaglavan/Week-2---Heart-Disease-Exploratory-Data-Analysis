*TODO 2 - Write code to be able to validate the following insight*

***Men are more represented in the dataset (around 68%), and more men than women are diagnosed.***

```python
# First show the actual number of men and women
gender_counts = df["sex"].value_counts()
print(f"Number of Men and Number of Women:")
print(f"Men: {gender_counts[1]}")
print(f"Women: {gender_counts[0]}")

# Show the percentage of men and women
total_patients = len(df)
men_percent = gender_counts[1]/total_patients
women_percent = gender_counts[0]/total_patients

# Now print the percentages
print()
print(f"Percentage of Men in the dataset: {men_percent*100:.1f}%")
print(f"Percentage of Women in the dataset: {women_percent*100:.1f}%")

# Now show that more Men are diagnosed then Women
men_diagnosed = (df[(df['sex'] == 1) & (df['target'] == 1)].shape[0])/gender_counts[1]
women_diagnosed = df[(df['sex'] == 0) & (df['target'] == 1)].shape[0]/gender_counts[0]

print()
print(f"Percentage of Men being diagnosed (Among Men): {men_diagnosed*100:.1f}%")
print(f"Percentage of Women being diagnosed (Among Women): {women_diagnosed*100:.1f}%")
```

*TODO 3 - Write code to be able to validate the following insight*

***Women with abnormal cholesterol or ECG results also show high prevalence.***

```python
# Collection of Data for the high cholesteral aspect
women_abnormal_chol = df[(df['sex'] == 0) & (df['chol'] >= 200)]

cholesterol_prevalence = (
    women_abnormal_chol[women_abnormal_chol['target'] == 1].shape[0]
    / women_abnormal_chol.shape[0]
)

print(f"Heart disease prevalence among women with abnormal cholesterol:\n{cholesterol_prevalence*100:.1f}%")

# Collection of Data for the ECG results
women_abnormal_ecg = df[ (df['sex'] == 0) & (df['restecg'].isin([1, 2]))]

ecg_prevalence = (
    women_abnormal_ecg[women_abnormal_ecg['target'] == 1].shape[0]
    / women_abnormal_ecg.shape[0]
)

print()
print(f"Heart disease prevalence among women with abnormal ECG:\n{ecg_prevalence*100:.1f}%")


# Now for women who have both
women_both_abnormal = df[(df['sex'] == 0) & (df['chol'] >= 200) & (df['restecg'].isin([1, 2]))]

both_prevalence = (
    women_both_abnormal[women_both_abnormal['target'] == 1].shape[0]
    / women_both_abnormal.shape[0]
)

print()
print(f"Heart disease prevalence among women with both abnormal cholesterol and ECG:\n{both_prevalence*100:.1f}%")

women_normal = df[(df['sex'] == 0) & (df['chol'] < 200) & (df['restecg'] == 0)]

normal_prevalence = (
    women_normal[women_normal['target'] == 1].shape[0]
    / women_normal.shape[0]
)

print()
print(f"Heart disease prevalence among women with neither abnormal result:\n{normal_prevalence*100:.1f}%")
```

*TODO 4 - Write code to be able to validate the following insight*

***Higher cholesterol and blood pressure are more common in patients with heart disease.***

```python
# Number of patients that have high cholesterol and blood pressure with heart disease
with_heart_disease = df[(df['chol'] >= 200) & (df['trestbps'] > 120) & (df['target'] == 1)]

# Number of pateients that have high cholesterol and blood pressure without heart disease
without_heart_disease = df[(df['chol'] >= 200) & (df['trestbps'] > 120) & (df['target'] == 0)]

# Now get the percentages
heart_disease_perc = len(with_heart_disease) / len(df)
without_perc = len(without_heart_disease) / len(df)

# Print percentages
print(f"Percentage of patients with high cholesterol and blood pressure as well with heart disease")
print(f"{heart_disease_perc*100:.1f}%")

print()
print(f"Percentage of patients with high cholesterol and blood pressure without heart disease")
print(f"{without_perc*100:.1f}%")

# Doing a scatter plot to see if there is a correlation between the two
plt.scatter(df['chol'], df['trestbps'], alpha=0.5, color = 'red')

plt.xlabel('Cholesterol')
plt.ylabel('Resting Blood Pressure')
plt.title('Cholesterol vs Resting Blood Pressure')

plt.show()

# Checking the correlation coefficiant between cholesterol and Resting Blood Pressure
correlation = df['chol'].corr(df['trestbps'])

# Printing it
print(f"\nCorrelation Coefficiant of cholesterol and resting blood pressure:")
print(f"{correlation:.2f}")
```

Since the correlation between cholesterol and resting blood pressure is so low, cholesterol and blood pressure should be compared seperatley

```python
# Cholesterol comparison

# Mean cholesterol for heart disease
chol_with = df[df['target'] == 1]['chol'].mean()
chol_without = df[df['target'] == 0]['chol'].mean()

# Print the difference
print(f"Average cholesterol of patients with heart disease:")
print(f"{chol_with:.2f}")

print()
print(f"Average cholesterol of patients without heart disease:")
print(f"{chol_without:.2f}")

# Resting blood pressure comparison
rbp_with = df[df['target'] == 1]['trestbps'].mean()
rbp_without = df[df['target'] == 0]['trestbps'].mean()

# Print the difference
print(f"\nAverage resting blood pressure of patients with heart disease:")
print(f"{rbp_with:.2f}")

print()
print(f"Average resting blood pressure of patients without heart disease:")
print(f"{rbp_without:.2f}")
```

```python
# The following is a histogram comparing the concentrations of cholesterol
# Among patients with heart disease and without

plt.hist(
    df[df['target'] == 1]['chol'],
    bins=10,
    alpha=0.5,
    label='Heart Disease',
    color = 'darkblue'
)

plt.hist(
    df[df['target'] == 0]['chol'],
    bins=10,
    alpha=0.5,
    label='No Heart Disease',
    color = 'yellow'
)

plt.xlabel('Cholesterol')
plt.ylabel('Frequency')
plt.title('Cholesterol Distribution by Heart Disease Status')
plt.legend()

plt.show()
```

```python
# The following is a histogram comparing the concentrations of resting
# Blood pressure among patients with and without heart disease
plt.hist(
    df[df['target'] == 1]['trestbps'],
    bins=10,
    alpha=0.5,
    density=True,
    label='Heart Disease',
    color = 'darkblue'
)

plt.hist(
    df[df['target'] == 0]['trestbps'],
    bins=10,
    alpha=0.5,
    density=True,
    label='No Heart Disease',
    color = 'yellow'
)

plt.xlabel('Resting Blood Pressure')
plt.ylabel('Density')
plt.title('Resting Blood Pressure Distribution by Heart Disease Status')
plt.legend()

plt.show()
```

***Patients with disease achieve lower maximum heart rates.***

```python
# Compare the means
mean_hr_hd = df[df['target'] == 1]['thalach'].mean()
mean_hr_no_hd = df[df['target'] == 0]['thalach'].mean()

print(f"Average max heart rate with heart disease:")
print(f"{mean_hr_hd:.2f}")

print()
print(f"Average max heart rate without heart disease:")
print(f"{mean_hr_no_hd:.2f}")

# Creating a box plot to show the difference in the medians visually
# To see that lower maximum heart rates are lower for patients with heart disease
sns.boxplot(
    data=df,
    x='target',
    y='thalach',
    hue = 'target',
    legend = False,
    palette=['red','lightblue']
)

plt.xlabel('Heart Disease Status')
plt.ylabel('Maximum Heart Rate')
plt.title('Maximum Heart Rate by Heart Disease Status')

plt.xticks([0, 1], ['No Heart Disease', 'Heart Disease'])

plt.show()
```

*TODO 6 - Come up with at least one additional insight to validate*

***Patients that are classified as asymptomatic have a higher heart disease rate than patients that experience chest pain.***


```python
# Getting percentages of those with heart disease and those without based
# On their chest pain levels

# Asymptomatic
asym_with = len(df[(df['cp'] == 3) & (df['target'] == 1)]) / len(df)
asym_without = len(df[(df['cp'] == 3) & (df['target'] == 0)]) / len(df)

# Not asymptomatic
cp_with = len(df[(df['cp'].isin([0,1,2])) & (df['target'] == 1)]) / len(df)
cp_without = len(df[(df['cp'].isin([0,1,2])) & (df['target'] == 0)]) / len(df)

# Print results
print(f"Percentage of patients that are asymptomatic with heart disease:")
print(f"{asym_with*100:.2f}%")

print()
print(f"Percentage of patients that are asymptomatic without heart disease:")
print(f"{asym_without*100:.2f}%")

print()
print(f"Percentage of patients with chest pain and heart disease:")
print(f"{cp_with*100:.2f}%")

print()
print(f"Percentage of patients with chest pain and without heart disease:")
print(f"{cp_without*100:.2f}%")

# Show totals
print(f"\nTOTAL DIAGNOSTIC RATES")
asym_total = asym_with + asym_without
cp_total = cp_with + cp_without

asym_diagnosis_rate = asym_with / asym_total
cp_diagnosis_rate = cp_with / cp_total

print()
print(f"Heart disease rate among asymptomatic patients:")
print(f"{asym_diagnosis_rate*100:.2f}%")

print()
print(f"Heart disease rate among patients with chest pain:")
print(f"{cp_diagnosis_rate*100:.2f}%")
```
